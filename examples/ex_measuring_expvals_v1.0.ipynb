```python
from qiskit.circuit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.primitives import StatevectorEstimator as Estimator
import numpy as np

# 1. Define the QuantumCircuit
qc = QuantumCircuit(3)
qc.ry(np.arccos(0.5), 0)
qc.ry(np.arccos(0.75), 1)
qc.ry(np.pi-np.arccos(0.5), 2)

# 2. Define the Observable (as per qubit ordering 0, 1, 2)
observable = [SparsePauliOp("IIZ"), SparsePauliOp("IZI"), SparsePauliOp("ZII")]

# 3. Instantiate the V2 Estimator
estimator = Estimator()

# 4. Run the job using the V2 signature
job = estimator.run([(qc, observable)])
result = job.result()

# The result contains an array of expectation values
exp_val = result[0].data.evs

print('Estimator "expval" example.')
print(f"Expectation value: {exp_val}")
```
